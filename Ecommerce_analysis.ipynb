__ORDERS__

```python
orders['order_year'] = orders['order_purchase_timestamp'].dt.year
orders['order_month'] = orders['order_purchase_timestamp'].dt.month
orders['month_name'] = orders['order_purchase_timestamp'].dt.month_name()
orders['day_of_week'] = orders['order_purchase_timestamp'].dt.day_name()
orders['quarter'] = 'Q' + orders['order_purchase_timestamp'].dt.quarter.astype(str)
orders['is_weekend'] = orders['order_purchase_timestamp'].dt.dayofweek>=5
orders['weekend_order'] = orders['is_weekend'].map({True:'Yes', False:'No'})
orders['delivery_days'] = (orders['order_delivered_customer_date']-orders['order_purchase_timestamp']).dt.days
orders['delivery_gap'] = (orders['order_estimated_delivery_date']-orders['order_delivered_customer_date']).dt.days
```

```python
def estimation(txt):
    if txt > 0:
        return 'Early'
    elif txt == 0:
        return 'On time'
    else:
        return 'Late'
```

```python
orders['delivery_gap'] = orders['delivery_gap'].apply(estimation)
```

```python
df = df.merge(orders[['order_id','order_year','order_month','month_name','day_of_week','quarter','is_weekend','weekend_order','delivery_days','delivery_gap']], on='order_id', how='left')
```

```python
df = df.merge(payments[['payment_type','payment_installments','order_id']], on='order_id', how='left')
```

```python
order_item_sum = order_items.groupby('order_id')[['price','freight_value']].sum().reset_index()
```

```python
df = df.merge(order_item_sum , on='order_id', how='left')
```

```python
df['total_charges'] = df['price_y']+ df['freight_value_y']
```

```python
tables = {
    'customers': customers,
    'sellers': sellers,
    'products':products,
    'categories': category,
    'geolocation': geolocation,
    'orders': orders,
    'orderItem': order_items,
    'payement': payments,
    'reviews': reviews,
}
```

```python
for table_name, df in tables.items():
    try:
        with engine.begin() as conn:
            conn.execute(text(f'TRUNCATE TABLE "{table_name}" RESTART IDENTITY CASCADE'))
            df.to_sql(
                name = table_name,
                con = conn,
                if_exists = 'append',
                index = False,
                method = 'multi'
            )
            print(f"{table_name} imported successfully with {len(df)} rows")
    except Exception as e:
           print(f"error in {table_name}: {e}")

print("Pipeline completed successfully")
```
